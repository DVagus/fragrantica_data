```python

import pandas as pd
```

```python
df = pd.read_csv("fragrantica.csv", encoding="iso-8859-1", sep=";") 
df.head()
```

```python
# replace commas with dots in 'Rating Value'
df['Rating Value'] = df['Rating Value'].str.replace(',', '.')

# convert 'Rating Value' from string to float and 'Rating Count' to integer
# use pd.to_numeric to convert strings to numbers
df["Rating Value"] = pd.to_numeric(df["Rating Value"])
df["Rating Count"] = pd.to_numeric(df["Rating Count"])
```

```python
# count total # of parfums
print("# parfums:", len(df))

# count total parfums with more than 50 Rating Count
print("# parfums with Rating Count >= 50:", len(df[df["Rating Count"] >= 50]))

# keep only parfums with Rating Count >= 50
df = df[df["Rating Count"] >= 50]
```

```python
# define the relevant columns to keep
columns_to_keep = ["Gender", "Rating Value", "mainaccord1", "mainaccord2", "mainaccord3"]

# keep only the specified columns
df = df[columns_to_keep]
```

```python
#save the new table perfumes.csv file
df.to_csv("parfums.csv", index=False, sep=",")
df = pd.read_csv("parfums.csv")
df.head()
```

```python
# missing values management 
# replace NaN in 'mainaccord2' and 'mainaccord3' with 'mainaccord1'
df["mainaccord2"] = df["mainaccord2"].fillna(df["mainaccord1"])
df["mainaccord3"] = df["mainaccord3"].fillna(df["mainaccord1"])
df.isna().sum()
```

```python
# create a binary column for success: 1 if Rating Value >= 4.0 (success), 0 otherwise
df["is_high_rated"] = (df["Rating Value"] >= 4.0).astype(int)
df
```

```python
def group_accord(accord):
    # Convert to lowercase to handle variations
    accord = str(accord).lower()
    
    # Citrus (Hespéridée) 
    if any(keyword in accord for keyword in [ 'citrus', 'fresh', 'ozonic', 'aquatic', 'marine', 'champagne', 'coca-cola', 'salty']):
        return 'Citrus'
    
    # Floral (Fleuri) 
    elif any(keyword in accord for keyword in ['floral', 'rose', 'iris', 'violet', 'tuberose', 'white floral', 'yellow floral', 'powdery', 'aldehydic']):
        return 'Floral'
    
    # Woody (Boisé)
    elif any(keyword in accord for keyword in ['woody', 'oud', 'patchouli', 'earthy', 'mossy', 'conifer', 'paper']):
        return 'Woody'
    
    # Fougère - Add herbal and aromatic for more examples
    elif any(keyword in accord for keyword in ['green','lavender', 'herbal', 'fresh spicy', 'camphor', 'cannabis']):
        return 'Fougère'
    
    # Chypre - Add resinous and earthy
    elif any(keyword in accord for keyword in ['mossy', 'earthy', 'mineral', 'clay', 'sand']):
        return 'Chypre'
    
    # Oriental - Keep sweet fruits and gourmand
    elif any(keyword in accord for keyword in ['vanilla', 'amber', 'musky', 'balsamic', 'sweet', 'gourmand', 'caramel', 'chocolate', 'coconut', 'honey', 'rum', 'whiskey', 'wine', 'tobacco', 'leather', 'smoky', 'spicy', 'warm spicy', 'soft spicy', 'anis', 'cinnamon', 'nutty', 'almond', 'beeswax', 'oily', 'lactonic', 'tropical', 'fruity', 'cherry', 'cacao', 'coffee', 'savory', 'animalic']):
        return 'Oriental'
    
    # Aromatic - Only for specific fresh/aquatic notes
    else:  # Includes green, fresh, aquatic, ozonic
        return 'Aromatic'

# grouping
df['accord1_group'] = df['mainaccord1'].apply(group_accord)
df['accord2_group'] = df['mainaccord2'].apply(group_accord)
df['accord3_group'] = df['mainaccord3'].apply(group_accord)

# Show distribution
'''
print("\nDistribuição de accord1_group:")
print(df['accord1_group'].value_counts())
print("\nDistribuição de accord2_group:")
print(df['accord2_group'].value_counts())
print("\nDistribuição de accord3_group:")
print(df['accord3_group'].value_counts())
'''
```
