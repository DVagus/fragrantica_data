```python

import pandas as pd
```

```python
df = pd.read_csv("fragrantica.csv", encoding="iso-8859-1", sep=";") 
df.head()
```

```python
# define the columns to keep
columns_to_keep = ["Gender", "Rating Value", "mainaccord1", "mainaccord2", "mainaccord3"]

# keep only the specified columns
df = df[columns_to_keep]
```

```python
df.head()
```

```python
# replace commas with dots in 'Rating Value'
df['Rating Value'] = df['Rating Value'].str.replace(',', '.')

# convert 'Rating Value' from string to float
# use pd.to_numeric to convert strings to numbers
df["Rating Value"] = pd.to_numeric(df["Rating Value"])
df.head()
```

```python
#count "NaN" values 
df.isna().sum()
```

```python
#save the new table perfumes.csv file
df.to_csv("parfums.csv", index=False, sep=',')
df = pd.read_csv("parfums.csv")
df.head()
```

```python
df.count()
```

```python
#count "NaN" values 
df.isna().sum()
```

```python
# replace NaN in 'mainaccord2' and 'mainaccord3' with 'mainaccord1'
df["mainaccord2"] = df["mainaccord2"].fillna(df["mainaccord1"])
df["mainaccord3"] = df["mainaccord3"].fillna(df["mainaccord1"])
df.isna().sum()
```

```python
# create a binary column for success: 1 if 'Rating Value' >= 4.0 (success), 0 otherwise
df["is_high_rated"] = (df["Rating Value"] >= 4.0).astype(int)
df
```
